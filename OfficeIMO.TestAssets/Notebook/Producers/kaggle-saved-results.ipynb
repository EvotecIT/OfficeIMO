Computational Times of Complexities per layer

Attention: O(n^2*d)

Reccurent: O(n*d^2)

n=sequence_length

d=token_dimension

```python
import numpy as np
import time
```

```python
n=512
d=512
# This means O(n^2*d)==O(n*d^2)
```

```python
n**2*d==n*d**2
```

```python
n*d**2
```

```python
input_seq=np.random.rand(n,d)
```

```python
input_seq.shape
```

# CPU

## Attention

```python
start_time=time.time()
for i in range(n):
    for j in range(n):
        res=np.dot(input_seq[i],input_seq[j])

estimated_time=time.time()-start_time
print(f"Self-Attention layer Computation Time: {estimated_time} Seconds.")
```

## Recurrent

```python
start_time=time.time()
hidden_state=np.zeros(d)
for i in range(n):
    for j in range(d):
        for k in range(d):
            hidden_state[j]+=input_seq[i,j]*hidden_state[k]

estimated_time=time.time()-start_time
```

```python
print(f"Recurrent layer Computation Time: {estimated_time} Seconds.")
```

```python
import torch
import time
```

```python
n=512
d=512
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)
```

```python
input_seq=torch.rand(n,d,device=device)
```

## Attention

```python
start_time = time.time()
_ = torch.mm(input_seq, input_seq.t())
estimated_time = time.time() - start_time
print(f"Self-attention computation time: {estimated_time} seconds")
```

## Recurrent

```python
start_time = time.time()
hidden_state = torch.zeros(d, device=device)
for i in range(n):
    for j in range(d):
        for k in range(d):
            hidden_state[j] += input_seq[i, j] * hidden_state[k]
            ct = time.time() - start_time
            if ct>estimated_time*10:
              break

estimated_time = time.time() - start_time
print(f"Recurrent layer computation time: {estimated_time} seconds")
```
